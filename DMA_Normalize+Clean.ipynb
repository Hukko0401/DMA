```python
dir_map = {
    "Đông - Nam": "Đông Nam",
    "Đông - Bắc": "Đông Bắc",
    "Tây - Nam": "Tây Nam",
    "Tây - Bắc": "Tây Bắc"
}

merged_df["direction"] = merged_df["direction"].replace(dir_map)
merged_df["balcony_direction"] = merged_df["balcony_direction"].replace(dir_map)
```

```python
def map_furnishing(x):

    if pd.isna(x) or x == "nan":
        return None

    if any(k in x for k in [
        "đầy đủ", "full", "đủ nội thất", "nội thất đầy đủ"
    ]):
        return "Đầy đủ"

    elif any(k in x for k in [
        "cơ bản", "ntcb", "nội thất cơ bản"
    ]):
        return "Cơ bản"

    elif any(k in x for k in [
        "không nội thất", "bàn giao thô", "nhà trống"
    ]):
        return "Không nội thất"

    else:
        return "Cơ bản"
```

```python
merged_df["furnishing_clean"] = merged_df["furnishing"].apply(map_furnishing)
```

```python
import pandas as pd

def map_legal_status(x):

    if pd.isna(x) or x == "nan":
        return None   # giữ NaN

    if any(k in x for k in [
        "sổ đỏ",
        "sổ hồng",
        "đã có sổ"
    ]):
        return "Sổ đỏ / Sổ hồng"

    elif any(k in x for k in [
        "hợp đồng mua bán",
        "hdmb",
        "hđmb"
    ]):
        return "Hợp đồng mua bán"

    elif any(k in x for k in [
        "chờ sổ",
        "đang chờ sổ",
        "chưa ra sổ"
    ]):
        return "Đang chờ sổ"

    else:
        return "Không rõ"
```

```python
merged_df["legal_status"] = merged_df["legal_status"].apply(map_legal_status)
```

```python
import pandas as pd

def map_ownership(x):

    if pd.isna(x) or x == "nan":
        return None

    if any(k in x for k in [
        "lâu dài",
        "không thời hạn",
        "sổ hồng",
        "sổ đỏ"
    ]):
        return "Lâu dài"

    elif any(k in x for k in [
        "chưa cấp sổ",
        "chưa có sổ"
    ]):
        return "Chưa có sổ"

    else:
        return "Không rõ"
```

```python
import pandas as pd
import unicodedata

# ===== NORMALIZE TEXT =====
def remove_accent(text):
    text = unicodedata.normalize('NFD', str(text))
    return ''.join(c for c in text if unicodedata.category(c) != 'Mn')

# ===== CITY =====
def normalize_city(city):
    if pd.isna(city):
        return "unknown"

    city = str(city).lower().strip()
    city = remove_accent(city)

    if any(x in city for x in ["hcm", "ho chi minh", "tphcm", "tp hcm", "sai gon"]):
        return "hcm"

    if any(x in city for x in ["ha noi", "hanoi", "hn"]):
        return "hn"

    if any(x in city for x in ["da nang", "danang"]):
        return "danang"

    return city


# ===== BRAND =====
def normalize_brand(brand):
    if pd.isna(brand):
        return "unknown"

    brand = str(brand).lower()

    if "vin" in brand:
        return "vinhomes"

    return brand


# ===== APPLY =====
merged_df = merged_df.copy()

merged_df["city"] = merged_df["city"].apply(normalize_city)
merged_df["brand"] = merged_df["brand"].apply(normalize_brand)

# ===== FIX PRICE =====
merged_df["price_vnd"] = merged_df["price_vnd"].fillna(0).astype(float).round().astype("int64")

# ===== OTHER NUMERIC =====
merged_df["area_m2"] = merged_df["area_m2"].astype(float)
merged_df["bedrooms"] = merged_df["bedrooms"].fillna(0).astype(int)
merged_df["bathrooms"] = merged_df["bathrooms"].fillna(0).astype(int)

# ===== OPTIONAL =====
if "project_name" in merged_df.columns:
    merged_df["project_name"] = merged_df["project_name"].str.lower().str.strip()

# ===== CHECK =====
print(merged_df[["city", "brand"]].drop_duplicates().head())
print(merged_df.dtypes)
```
