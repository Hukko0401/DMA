```python
selected_cols = [
    # numeric
    'area_m2', 'bedrooms', 'bathrooms',
    'dist_cbd_km', 'dist_metro_km', 'dist_river_lake_km',
    'dist_highway_km', 'dist_airport_km',
    'amenity_score', 'description_length',
    'total_apartment', 'construction_density',

    # binary
    'Metro_city','luxury', 'river_view', 'investment',
    'modern_urban', 'full_furniture', 'high_floor',

    # categorical
    'city', 'brand','title','description'
]

df = df[selected_cols + ['price_vnd']].copy()
```

```python
import numpy as np

df['log_price'] = np.log(df['price_vnd'])
df = df.drop(columns=['price_vnd'])
```

```python
# flag
df['bed_missing'] = df['bedrooms'].isnull().astype(int)
df['bath_missing'] = df['bathrooms'].isnull().astype(int)

# fill median
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].median())

# dist metro
max_dist = df['dist_metro_km'].max()
df['dist_metro_km'] = df['dist_metro_km'].fillna(max_dist + 1)
```

```python
def normalize_city(x):
    x = str(x).lower()
    if 'hcm' in x or 'ho chi minh' in x:
        return 'hcm'
    elif 'ha noi' in x or 'hanoi' in x:
        return 'hanoi'
    else:
        return x

df['city'] = df['city'].apply(normalize_city)
```

```python
df['total_apartment'] = np.log1p(df['total_apartment'])
df['description_length'] = np.log1p(df['description_length'])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

def evaluate_model(model, X_train, y_train, X_val, y_val, X_test, y_test, model_name="Model"):
    model.fit(X_train, y_train)

    # predictions
    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)
    y_test_pred = model.predict(X_test)

    # metrics function
    def get_metrics(y_true, y_pred):
        return {
            "MAE": mean_absolute_error(y_true, y_pred),
            "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
            "R2": r2_score(y_true, y_pred)
        }

    results = {
        "Train": get_metrics(y_train, y_train_pred),
        "Val": get_metrics(y_val, y_val_pred),
        "Test": get_metrics(y_test, y_test_pred)
    }

    print(f"\n===== {model_name} =====")
    for split, metric in results.items():
        print(f"{split}: {metric}")

    return results
```

```python
def evaluate_all(models_dict, X_train, y_train, X_val, y_val, X_test, y_test):
    all_results = {}

    for name, model in models_dict.items():
        results = evaluate_model(
            model,
            X_train, y_train,
            X_val, y_val,
            X_test, y_test,
            model_name=name
        )
        all_results[name] = results

    return all_results
```

```python
import joblib
import numpy as np
import pandas as pd
from fastapi import FastAPI
from scipy.sparse import hstack

import nest_asyncio
from pyngrok import ngrok
import uvicorn

# ===== LOAD =====
model = joblib.load("model.pkl")
scaler = joblib.load("scaler.pkl")
tfidf = joblib.load("tfidf.pkl")
columns = joblib.load("columns.pkl")

app = FastAPI()

# ===== PREPROCESS =====
def preprocess(data: dict):

    import pandas as pd
    import numpy as np
    from scipy.sparse import hstack

    df = pd.DataFrame([data])

    # ===== TEXT =====
    df["title"] = df.get("title", "")
    df["description"] = df.get("description", "")
    df["text"] = df["title"] + " " + df["description"]

    # ===== CITY NORMALIZE =====
    df["city"] = df.get("city", "hcm")
    df["city"] = df["city"].apply(
        lambda x: "hcm" if "hcm" in str(x).lower() else (
            "hanoi" if "ha noi" in str(x).lower() else x
        )
    )

    # ===== NUMERIC FEATURES =====
    num_cols = [
        'area_m2', 'bedrooms', 'bathrooms',
        'dist_cbd_km', 'dist_metro_km', 'dist_river_lake_km',
        'dist_highway_km', 'dist_airport_km',
        'amenity_score', 'description_length',
        'total_apartment', 'construction_density'
    ]

    for col in num_cols:
        df[col] = df.get(col, 0)

    # ===== BINARY FEATURES =====
    binary_cols = [
        'Metro_city','luxury','river_view','investment',
        'modern_urban','full_furniture','high_floor'
    ]

    for col in binary_cols:
        df[col] = df.get(col, 0)

    # ===== LOG TRANSFORM (GIỐNG TRAIN) =====
    df['total_apartment'] = np.log1p(df['total_apartment'])
    df['description_length'] = np.log1p(df['description_length'])

    # ===== ONE HOT =====
    text_data = df["text"].copy()
    df = pd.get_dummies(df)

    # ===== ALIGN COLUMN =====
    for col in columns:
        if col not in df:
            df[col] = 0

    df = df[columns]

    # ===== SCALE ONLY NUMERIC =====
    df[num_cols] = scaler.transform(df[num_cols])

    # ===== TO NUMPY =====
    df_np = df.astype(float).to_numpy()

    # ===== TFIDF =====
    text_vec = tfidf.transform(text_data)

    # ===== COMBINE =====
    final = hstack([df_np, text_vec])

    return final


# ===== API =====
@app.post("/predict")
def predict(data: dict):
    try:
        print("📥 INPUT:", data)

        X = preprocess(data)
        print("✅ Preprocess OK")

        pred_log = model.predict(X)[0]
        print("✅ Model OK")

        pred_price = np.exp(pred_log)
        print("💰 PRICE:", pred_price)

        return {
            "predicted_price": float(pred_price)
        }

    except Exception as e:
        import traceback
        print("🔥 ERROR OCCURRED:")
        traceback.print_exc()

        return {
            "error": str(e)
        }

# ===== RUN =====
nest_asyncio.apply()

from threading import Thread
import time

# ===== START SERVER TRƯỚC =====
def run():
    uvicorn.run(app, host="0.0.0.0", port=8000)

thread = Thread(target=run)
thread.start()

# ===== ĐỢI SERVER KHỞI ĐỘNG =====
time.sleep(2)

# ===== START NGROK SAU =====
ngrok.set_auth_token("3BUdzh3ONBVlqVCcT0KZmFo0NhN_71NNpDNuxrSqXYLx7bJgN")

public_url = ngrok.connect(8000)

print("Public URL:", public_url)
print("🌐 BASE URL:", public_url.public_url)
print("📄 DOCS:", public_url.public_url + "/docs")
print("🚀 API:", public_url.public_url + "/predict")
```
